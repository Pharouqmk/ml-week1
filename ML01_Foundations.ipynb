```python
# loading dataset in to pandas
import pandas as pd

customers = pd.read_csv('ALA_ShopSmart_Customers.csv')
employees = pd.read_csv('ShopSmart_Employees.csv')
inventory = pd.read_csv('ShopSmart_Inventory.csv')
orders = pd.read_csv('ShopSmart_Orders.csv')
order_items = pd.read_csv('ShopSmart_Order_Items.csv')
products = pd.read_csv('ShopSmart_Products.csv')
returns = pd.read_csv('ShopSmart_Returns.csv')
stores = pd.read_csv('ShopSmart_Stores.csv')
suppliers = pd.read_csv('ShopSmart_Suppliers.csv')

print("All datasets loaded successfully")
```

```python
# checking thougth each data set
print(orders.shape)
print(orders.columns.tolist())
print(orders.dtypes)
print(orders.isnull().sum())
orders.head()
```

```python
# fixing date column and checking for duplicate rows
print("Duplicate rows:", orders.duplicated().sum())

orders['Order_Date'] = pd.to_datetime(orders['Order_Date'])
print(orders['Order_Date'].dtype)
print(orders['Order_Date'].min(), "to", orders['Order_Date'].max())
```

```python
print(order_items.shape)
print(order_items.dtypes)
print(order_items.isnull().sum())
print("Duplicates:", order_items.duplicated().sum())
order_items.head()
```

```python
# for product
print(products.shape)
print(products.dtypes)
print(products.isnull().sum())
print("Duplicates:", products.duplicated().sum())
products.head()
```

```python
# customers
print(customers.shape)
print(customers.dtypes)
print(customers.isnull().sum())
print("Duplicates:", customers.duplicated().sum())
customers.head()
```

```python
# checking for duplicate rows in (stores, suppliers, inventory, returns) using function
for name, df in [('stores', stores), ('suppliers', suppliers), ('inventory', inventory), ('returns', returns)]:
    print(f"=== {name} ===")
    print("Shape:", df.shape)
    print("Missing values:\n", df.isnull().sum())
    print("Duplicates:", df.duplicated().sum())
    print()
```

```python
# checking for rlational data
print(orders['Order_ID'].nunique(), "unique orders")
print(order_items['Order_ID'].nunique(), "unique order_ids in order_items")
print(len(order_items), "total order_item rows")
```

```python
# marging order with order_item
master = orders.merge(order_items, on='Order_ID', suffixes=('', '_item'))
print(master.shape)
master.head()
```

```python
# verifying for identical, then drop the duplicates:
print((master['Product_ID'] == master['Product_ID_item']).all())
print((master['Quantity'] == master['Quantity_item']).all())
print((master['Unit_Price'] == master['Unit_Price_item']).all())
print((master['Discount'] == master['Discount_item']).all())
```

```python
# cleaning
master = master.drop(columns=['Order_Item_ID', 'Product_ID_item', 'Quantity_item', 'Unit_Price_item', 'Discount_item'])
print(master.shape)
master.columns.tolist()
```

```python
# adding products, customers, and stores
master = master.merge(products, on='Product_ID')
master = master.merge(customers, on='Customer_ID')
master = master.merge(stores, on='Store_ID')
print(master.shape)
master.columns.tolist()
```

```python
# Ambiguous column names
master = master.rename(columns={
    'City_x': 'Customer_City',
    'State_x': 'Customer_State',
    'City_y': 'Store_City',
    'State_y': 'Store_State'
})
```

```python
# Creating the target variable
master['Order_Total'] = master['Line_Total'] + master['Shipping_Cost']
```

```python
# Return flag for calassification
master['Return_Flag'] = master['Order_ID'].isin(returns['Order_ID']).astype(int)
print(master['Return_Flag'].value_counts())
```

```python
# future engineering
master['Order_Month'] = master['Order_Date'].dt.month
master['Order_Quarter'] = master['Order_Date'].dt.quarter
master['Profit_Margin'] = master['Selling_Price'] - master['Cost_Price']
master['Discount_Percentage'] = (master['Discount'] / master['Unit_Price']) * 100

print(master[['Order_Month', 'Order_Quarter', 'Profit_Margin', 'Discount_Percentage']].describe())
```

```python
# adding features for Customer Age Group, Customer Lifetime Value, and Days Since Last Purchase
master['Customer_Age_Group'] = pd.cut(master['Age'],
    bins=[0, 25, 35, 45, 55, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56+'])

clv = master.groupby('Customer_ID')['Order_Total'].sum().rename('Customer_Lifetime_Value')
master = master.merge(clv, on='Customer_ID')

reference_date = master['Order_Date'].max()
last_purchase = master.groupby('Customer_ID')['Order_Date'].max().rename('Last_Purchase_Date')
master = master.merge(last_purchase, on='Customer_ID')
master['Days_Since_Last_Purchase'] = (reference_date - master['Last_Purchase_Date']).dt.days

print(master['Customer_Age_Group'].value_counts())
print(master[['Customer_Lifetime_Value', 'Days_Since_Last_Purchase']].describe())
```

```python
# Adding inventory features
master = master.merge(
    inventory[['Product_ID', 'Store_ID', 'Stock_On_Hand', 'Reorder_Level']],
    on=['Product_ID', 'Store_ID'],
    how='left'
)
print(master.shape)
print("Missing stock info:", master['Stock_On_Hand'].isnull().sum())
```

```python
# investigating
print("Inventory rows:", inventory.shape[0])
print("Unique Product_IDs in inventory:", inventory['Product_ID'].nunique())
print("Unique Store_IDs in inventory:", inventory['Store_ID'].nunique())
print("Unique Product-Store combos in inventory:", inventory.drop_duplicates(['Product_ID','Store_ID']).shape[0])
```

```python
# Dropping the failed merge and redo it, joining on Product_ID only
master = master.drop(columns=['Stock_On_Hand', 'Reorder_Level'])

master = master.merge(
    inventory[['Product_ID', 'Stock_On_Hand', 'Reorder_Level']],
    on='Product_ID',
    how='left'
)
print(master.shape)
print("Missing stock info:", master['Stock_On_Hand'].isnull().sum())
```

```python
# intentory status
master['Inventory_Status'] = pd.cut(master['Stock_On_Hand'],
    bins=[-1, 20, 100, 1000],
    labels=['Low', 'Medium', 'High'])

print(master.shape)
print(master['Inventory_Status'].value_counts())
```

```python
# checking for usefull model input
print(master.dtypes)
```

```python
# building the clean feature set
drop_cols = ['Order_ID', 'Customer_ID', 'Product_ID', 'Store_ID',
             'Product_Name', 'Store_Name', 'First_Name', 'Last_Name', 'Email', 'Phone',
             'Order_Date', 'Registration_Date', 'Opening_Date', 'Last_Purchase_Date',
             'Sub_Category', 'Brand', 'Line_Total', 'Reorder_Level']

model_data = master.drop(columns=drop_cols)
print(model_data.shape)
print(model_data.columns.tolist())
```

```python
# encoding and scaling
for col in ['Payment_Method', 'Sales_Channel', 'Order_Status', 'Category', 'Gender',
            'Customer_City', 'Customer_State', 'Customer_Segment', 'Region',
            'Store_State', 'Store_City', 'Customer_Age_Group', 'Inventory_Status']:
    print(f"{col}: {model_data[col].nunique()} unique values")
```

```python
# One-hot encode all categorical columns:
categorical_cols = ['Payment_Method', 'Sales_Channel', 'Order_Status', 'Category', 'Gender',
                     'Customer_City', 'Customer_State', 'Customer_Segment', 'Region',
                     'Store_State', 'Store_City', 'Customer_Age_Group', 'Inventory_Status']

model_data_encoded = pd.get_dummies(model_data, columns=categorical_cols, drop_first=True)
print(model_data_encoded.shape)
```

```python
# Creating separate X (features) and y (targets) for each model, dropping the other target to prevent leakage
# For regression (predicting Order_Total)
X_reg = model_data_encoded.drop(columns=['Order_Total', 'Return_Flag'])
y_reg = model_data_encoded['Order_Total']

# For classification (predicting Return_Flag)
X_clf = model_data_encoded.drop(columns=['Order_Total', 'Return_Flag'])
y_clf = model_data_encoded['Return_Flag']

print("Regression features:", X_reg.shape)
print("Classification features:", X_clf.shape)
```

```python
# Split into training and test sets
from sklearn.model_selection import train_test_split

X_train, X_test, y_reg_train, y_reg_test, y_clf_train, y_clf_test = train_test_split(
    X_reg, y_reg, y_clf, test_size=0.2, random_state=42, stratify=y_clf
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
print("Train return rate:", y_clf_train.mean())
print("Test return rate:", y_clf_test.mean())
```

```python
# confirm the shapes look right:
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("Train return rate:", y_clf_train.mean())
print("Test return rate:", y_clf_test.mean())
```

```python
# Runnig the scaling code
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Scaling complete:", X_train_scaled.shape, X_test_scaled.shape)
```

```python
# Build your first Regression Model (predicting Order_Total). Step 1: Train a Linear Regression model.
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()
lr_model.fit(X_train_scaled, y_reg_train)

y_reg_pred = lr_model.predict(X_test_scaled)

print("Model trained successfully")
```

```python
# Evaluating it with MAE, RMSE, and R² Score — the three metrics for brief :
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_reg_test, y_reg_pred)
rmse = np.sqrt(mean_squared_error(y_reg_test, y_reg_pred))
r2 = r2_score(y_reg_test, y_reg_pred)

print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R² Score: {r2:.4f}")
```

```python
# testing
print("Average Order_Total:", y_reg_test.mean())
print("MAE as % of average order:", (mae / y_reg_test.mean()) * 100)
```

```python
# visualizing it — Actual vs Predicted, and the Residual Plot
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Actual vs Predicted
axes[0].scatter(y_reg_test, y_reg_pred, alpha=0.3, s=10)
axes[0].plot([y_reg_test.min(), y_reg_test.max()], [y_reg_test.min(), y_reg_test.max()], 'r--')
axes[0].set_xlabel('Actual Order_Total')
axes[0].set_ylabel('Predicted Order_Total')
axes[0].set_title('Actual vs Predicted')

# Residual Plot
residuals = y_reg_test - y_reg_pred
axes[1].scatter(y_reg_pred, residuals, alpha=0.3, s=10)
axes[1].axhline(y=0, color='r', linestyle='--')
axes[1].set_xlabel('Predicted Order_Total')
axes[1].set_ylabel('Residual (Actual - Predicted)')
axes[1].set_title('Residual Plot')

plt.tight_layout()
plt.savefig('regression_plots.png', dpi=150)
plt.show()
```

```python
# fixing with one targeted feature — an interaction term:
X_train_improved = X_train.copy()
X_test_improved = X_test.copy()

X_train_improved['Quantity_x_UnitPrice'] = X_train['Quantity'] * X_train['Unit_Price']
X_test_improved['Quantity_x_UnitPrice'] = X_test['Quantity'] * X_test['Unit_Price']

scaler2 = StandardScaler()
X_train_improved_scaled = scaler2.fit_transform(X_train_improved)
X_test_improved_scaled = scaler2.transform(X_test_improved)

lr_model2 = LinearRegression()
lr_model2.fit(X_train_improved_scaled, y_reg_train)
y_reg_pred2 = lr_model2.predict(X_test_improved_scaled)

mae2 = mean_absolute_error(y_reg_test, y_reg_pred2)
rmse2 = np.sqrt(mean_squared_error(y_reg_test, y_reg_pred2))
r2_2 = r2_score(y_reg_test, y_reg_pred2)

print(f"Improved MAE: {mae2:.2f} (was {mae:.2f})")
print(f"Improved RMSE: {rmse2:.2f} (was {rmse:.2f})")
print(f"Improved R²: {r2_2:.4f} (was {r2:.4f})")
```

```python
# removing the components that make Order_Total deterministic, and predict it from genuinely independent information instead
leak_cols = ['Quantity', 'Unit_Price', 'Discount', 'Shipping_Cost', 'Quantity_x_UnitPrice']

X_train_clean = X_train_improved.drop(columns=leak_cols)
X_test_clean = X_test_improved.drop(columns=leak_cols)

scaler3 = StandardScaler()
X_train_clean_scaled = scaler3.fit_transform(X_train_clean)
X_test_clean_scaled = scaler3.transform(X_test_clean)

lr_model3 = LinearRegression()
lr_model3.fit(X_train_clean_scaled, y_reg_train)
y_reg_pred3 = lr_model3.predict(X_test_clean_scaled)

mae3 = mean_absolute_error(y_reg_test, y_reg_pred3)
rmse3 = np.sqrt(mean_squared_error(y_reg_test, y_reg_pred3))
r2_3 = r2_score(y_reg_test, y_reg_pred3)

print(f"Leak-free MAE: {mae3:.2f}")
print(f"Leak-free RMSE: {rmse3:.2f}")
print(f"Leak-free R²: {r2_3:.4f}")
```

```python
# Training the Logistic Regression model (using our original, non-leaky scaled features from before):
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
log_model.fit(X_train_scaled, y_clf_train)

y_clf_pred = log_model.predict(X_test_scaled)

print("Model trained successfully")
```

```python
# Evaluating with Accuracy, Precision, Recall, F1, and Confusion Matrix:
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

accuracy = accuracy_score(y_clf_test, y_clf_pred)
precision = precision_score(y_clf_test, y_clf_pred)
recall = recall_score(y_clf_test, y_clf_pred)
f1 = f1_score(y_clf_test, y_clf_pred)

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1 Score: {f1:.4f}")
print("\nConfusion Matrix:")
print(confusion_matrix(y_clf_test, y_clf_pred))
print("\n", classification_report(y_clf_test, y_clf_pred))
```

```python
# Check what accuracy alone would have looked like without balancing
log_model_unbalanced = LogisticRegression(max_iter=1000, random_state=42)
log_model_unbalanced.fit(X_train_scaled, y_clf_train)
y_clf_pred_unbal = log_model_unbalanced.predict(X_test_scaled)

print("Unbalanced Accuracy:", accuracy_score(y_clf_test, y_clf_pred_unbal))
print("Unbalanced Recall:", recall_score(y_clf_test, y_clf_pred_unbal))
```
